```python
import subprocess

command = [
    "ffprobe",
    "-v", "error",
    "-show_entries",
    "format=duration",
    "-of",
    "default=noprint_wrappers=1:nokey=1",
    AUDIO_FILE
]

result = subprocess.run(
    command,
    capture_output=True,
    text=True
)

if result.returncode != 0:
    raise RuntimeError(
        f"Gagal membaca durasi audio:\n{result.stderr}"
    )

duration_seconds = float(
    result.stdout.strip()
)

hours = int(duration_seconds // 3600)
minutes = int((duration_seconds % 3600) // 60)
seconds = int(duration_seconds % 60)

print(
    f"Audio duration: "
    f"{hours:02d}:{minutes:02d}:{seconds:02d}"
)

print(
    f"Total seconds: "
    f"{duration_seconds:.2f}"
)
```

```python
import os
import glob
import site
import ctypes

# Programmatically locate cuBLAS and cuDNN and append to LD_LIBRARY_PATH
site_paths = site.getsitepackages()
cublas_files = []
cudnn_files = []

for path in site_paths:
    cublas_files += glob.glob(path + "/nvidia/cublas/**/libcublas.so.12", recursive=True)
    cudnn_files += glob.glob(path + "/nvidia/cudnn/**/libcudnn.so.9", recursive=True)

if cublas_files and cudnn_files:
    cublas_path = cublas_files[0]
    cudnn_path = cudnn_files[0]

    print("Loading cuBLAS from:", cublas_path)
    print("Loading cuDNN from:", cudnn_path)

    # Load libraries directly into the process memory using ctypes
    try:
        ctypes.CDLL(cublas_path, mode=ctypes.RTLD_GLOBAL)
        ctypes.CDLL(cudnn_path, mode=ctypes.RTLD_GLOBAL)
        print("✅ Libraries loaded successfully into memory!")
    except Exception as e:
        print(f"⚠️ Gagal memuat library menggunakan ctypes: {e}")

    # Update LD_LIBRARY_PATH environment variable as fallback
    cublas_dir = os.path.dirname(cublas_path)
    cudnn_dir = os.path.dirname(cudnn_path)
    existing_path = os.environ.get("LD_LIBRARY_PATH", "")
    new_path = f"{cublas_dir}:{cudnn_dir}:/usr/lib64-nvidia"
    if existing_path:
        new_path = f"{new_path}:{existing_path}"
    os.environ["LD_LIBRARY_PATH"] = new_path
else:
    print("⚠️ Could not automatically locate cuBLAS or cuDNN paths. Please ensure they are installed.")

print("=" * 60)
print("TEST TRANSCRIPTION")
print("=" * 60)

# Run the transcription
segments, info = model.transcribe(
    AUDIO_FILE,
    language="id",
    beam_size=10,
    vad_filter=True,
    initial_prompt=initial_prompt
)

print("Detected language:")
print(info.language)

print("\nLanguage probability:")
print(info.language_probability)

print("\nTranscription:")
for segment in segments:
    print(f"[{segment.start:.2f}s - {segment.end:.2f}s] {segment.text}")
```

```python
import time

print("=" * 60)
print("FULL TRANSCRIPTION")
print("=" * 60)

start_time = time.time()

segments, info = model.transcribe(
    AUDIO_FILE,
    language="id",
    beam_size=10,
    vad_filter=True,
    word_timestamps=False,
    initial_prompt=initial_prompt
)

print("Transcription started...")
print("Language:", info.language)
print(
    "Language probability:",
    info.language_probability
)
```

```python
transcription_segments = []

print("=" * 60)
print("PROCESSING SEGMENTS")
print("=" * 60)

for segment in segments:

    item = {
        "id": len(transcription_segments) + 1,
        "start": float(segment.start),
        "end": float(segment.end),
        "text": segment.text.strip()
    }

    transcription_segments.append(item)

print("\n✅ Transcription selesai.")
print(
    "Total segments:",
    len(transcription_segments)
)
```

```python
print("=" * 60)
print("TRANSCRIPTION PREVIEW")
print("=" * 60)

for segment in transcription_segments[:20]:

    print(
        f"[{segment['start']:.2f}s -> "
        f"{segment['end']:.2f}s]"
    )

    print(segment["text"])
    print()
```

```python
elapsed_time = time.time() - start_time

hours = int(elapsed_time // 3600)
minutes = int((elapsed_time % 3600) // 60)
seconds = int(elapsed_time % 60)

print("=" * 60)
print("PROCESSING TIME")
print("=" * 60)

print(
    f"{hours:02d}:"
    f"{minutes:02d}:"
    f"{seconds:02d}"
)

print(
    f"\nAudio duration: "
    f"{duration_seconds / 3600:.2f} hours"
)

print(
    f"Processing time: "
    f"{elapsed_time / 60:.2f} minutes"
)
```

```python
txt_path = os.path.join(
    OUTPUT_DIR,
    "transcript.txt"
)

with open(
    txt_path,
    "w",
    encoding="utf-8"
) as file:

    for segment in transcription_segments:

        file.write(
            f"[{segment['start']:.2f} - "
            f"{segment['end']:.2f}] "
            f"{segment['text']}\n"
        )

print("✅ TXT berhasil dibuat:")
print(txt_path)
```

```python
import json

json_path = os.path.join(
    OUTPUT_DIR,
    "transcript.json"
)

result_data = {
    "audio_file": AUDIO_FILENAME,
    "model": MODEL_NAME,
    "language": info.language,
    "language_probability": info.language_probability,
    "duration_seconds": duration_seconds,
    "processing_time_seconds": elapsed_time,
    "segments": transcription_segments
}

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        result_data,
        file,
        ensure_ascii=False,
        indent=2
    )

print("✅ JSON berhasil dibuat:")
print(json_path)
```

```python
from reportlab.lib.pagesizes import A4
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer
)
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.enums import TA_JUSTIFY
from reportlab.lib.units import cm
```

```python
pdf_path = os.path.join(
    OUTPUT_DIR,
    "transcript.pdf"
)

styles = getSampleStyleSheet()

body_style = styles["BodyText"]
body_style.fontName = "Helvetica"
body_style.fontSize = 11
body_style.leading = 17
body_style.alignment = TA_JUSTIFY
body_style.spaceAfter = 10

doc = SimpleDocTemplate(
    pdf_path,
    pagesize=A4,
    rightMargin=2.5 * cm,
    leftMargin=2.5 * cm,
    topMargin=2.5 * cm,
    bottomMargin=2.5 * cm
)

story = []

# Pecah menjadi beberapa paragraf berdasarkan panjang teks
paragraph_size = 1500

for i in range(
    0,
    len(full_text),
    paragraph_size
):
    paragraph_text = full_text[
        i:i + paragraph_size
    ]

    story.append(
        Paragraph(
            paragraph_text,
            body_style
        )
    )

    story.append(
        Spacer(1, 8)
    )

doc.build(story)

print("✅ PDF berhasil dibuat:")
print(pdf_path)
```
